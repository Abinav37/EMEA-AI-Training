```python

# Final Challenge: Create a separate cleaned DataFrame from the original data dictionary.
cleaned_df = pd.DataFrame(data).copy()

# Standardize text fields without modifying the original data dictionary.
cleaned_df["Name"] = cleaned_df["Name"].astype("string").str.strip().str.title()
cleaned_df["City"] = cleaned_df["City"].astype("string").str.strip().str.title()
cleaned_df["Department"] = cleaned_df["Department"].astype("string").str.strip().str.title()

cleaned_df["Gender"] = (
    cleaned_df["Gender"].astype("string").str.strip().str.lower().map({
        "m": "Male",
        "male": "Male",
        "f": "Female",
        "female": "Female"
    })
)

# Remove duplicate records.
cleaned_df = cleaned_df.drop_duplicates()

# Convert numeric columns and handle invalid values.
cleaned_df["Age"] = pd.to_numeric(cleaned_df["Age"], errors="coerce")
cleaned_df.loc[(cleaned_df["Age"] < 0) | (cleaned_df["Age"] > 100), "Age"] = np.nan

cleaned_df["Purchase_Amount"] = pd.to_numeric(cleaned_df["Purchase_Amount"], errors="coerce")
cleaned_df.loc[cleaned_df["Purchase_Amount"] < 0, "Purchase_Amount"] = np.nan

cleaned_df["Discount"] = pd.to_numeric(cleaned_df["Discount"], errors="coerce")

# Convert dates safely and derive year/month columns.
cleaned_df["Joining_Date"] = pd.to_datetime(cleaned_df["Joining_Date"], errors="coerce")
cleaned_df["Joining_Year"] = cleaned_df["Joining_Date"].dt.year
cleaned_df["Joining_Month"] = cleaned_df["Joining_Date"].dt.month

# Email validation and status.
cleaned_df["Email"] = cleaned_df["Email"].astype("string")
email_missing = cleaned_df["Email"].isna() | cleaned_df["Email"].str.strip().eq("")
email_valid = cleaned_df["Email"].fillna("").astype(str).str.contains(r"@.+\.[A-Za-z]{2,}", na=False)
cleaned_df["Email_Status"] = np.select([email_missing, email_valid], ["Missing", "Valid"], default="Invalid")

# Final summary metrics.
print("Number of records:", len(cleaned_df))
print("Number of columns:", cleaned_df.shape[1])
print("Missing values:\n", cleaned_df.isna().sum())
print("Duplicate records:", cleaned_df.duplicated().sum())
print("Average age:", cleaned_df["Age"].mean())
print("Average purchase amount:", cleaned_df["Purchase_Amount"].mean())
print("Maximum purchase amount:", cleaned_df["Purchase_Amount"].max())
print("Minimum purchase amount:", cleaned_df["Purchase_Amount"].min())
print("Number of cities:", cleaned_df["City"].nunique())
print("Number of departments:", cleaned_df["Department"].nunique())
print("Number of valid emails:", (cleaned_df["Email_Status"] == "Valid").sum())
print("Number of invalid emails:", (cleaned_df["Email_Status"] == "Invalid").sum())

cleaned_df
```
